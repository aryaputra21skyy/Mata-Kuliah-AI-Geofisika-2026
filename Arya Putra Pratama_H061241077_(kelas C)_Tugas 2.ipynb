## **1.  RTCIKOQV — Sistem Informasi Geografis (SIG)**

**R — Role**  
Bertindaklah sebagai seorang ahli Geofisika yang memiliki keahlian dalam Sistem Informasi Geografis (SIG), penginderaan jauh, analisis spasial, dan pemetaan geofisika. Anda memahami konsep geospasial secara teoritis sekaligus mampu menerapkannya dalam suatu permasalahan geofisika yang nyata.

**T — Task**  
Rancang sebuah topik penelitian/proyek SIG dalam bidang geofisika yang menarik, realistis, memiliki nilai ilmiah, dan dapat dikembangkan menjadi penelitian atau paper. Fokus utama harus berada pada pemanfaatan SIG untuk menyelesaikan suatu permasalahan geofisika atau kebumian, bukan sekadar membuat peta.

**C — Context**  
Gunakan konteks permasalahan geofisika yang relevan dengan kondisi wilayah Indonesia. Data yang digunakan dapat berasal dari DEM, data geologi, tutupan lahan, jaringan sungai, sesar, data penginderaan jauh, data seismik, maupun data spasial lainnya.

Contoh parameter yang dapat dipertimbangkan:
- elevasi dan kemiringan lereng,
- jarak terhadap sesar,
- jarak terhadap sungai,
- litologi/geologi,
- tutupan lahan,
- penggunaan lahan,
- atau parameter geofisika lain yang relevan.

Jika diperlukan, gunakan metode overlay, weighted overlay, reclassify, buffer, distance analysis, atau analisis spasial lainnya. Pemilihan metode harus disesuaikan dengan tujuan penelitian dan dijelaskan secara ilmiah.

**I — Instruction**  
Susun proyek secara sistematis mulai dari identifikasi masalah, tujuan, data yang diperlukan, metode analisis, tahapan pengolahan, hingga interpretasi hasil. Jelaskan hubungan antara setiap parameter dengan fenomena geofisika yang dikaji.

Alur penelitian secara umum:

**Data → Preprocessing → Analisis Spasial → Integrasi Parameter → Peta Tematik → Interpretasi Geofisika → Kesimpulan**

Jangan hanya menjelaskan bagaimana membuat peta. Tunjukkan pula mengapa analisis tersebut penting secara geofisika dan bagaimana hasilnya dapat digunakan untuk memahami kondisi suatu wilayah.

**K — Knowledge**  
Gunakan konsep dasar SIG dan geofisika secara benar, terutama mengenai:

- sistem koordinat dan proyeksi,
- analisis raster dan vektor,
- overlay spasial,
- klasifikasi dan pembobotan,
- analisis jarak,
- topografi,
- hubungan kondisi geologi dengan fenomena geofisika.

Apabila digunakan metode weighted overlay, jelaskan konsepnya dengan persamaan:

$$
S = \sum_{i=1}^{n} w_i x_i
$$

dengan $(S)$ sebagai skor akhir, $(w_i)$ sebagai bobot setiap parameter, dan $(x_i)$ sebagai skor masing-masing parameter.

Pemilihan bobot tidak boleh dibuat secara sembarangan. Berikan alasan ilmiah berdasarkan hubungan parameter terhadap fenomena yang diteliti.

**O — Output**  
Hasilkan rancangan penelitian SIG yang memiliki:

1. Judul penelitian yang menarik dan spesifik.
2. Latar belakang singkat.
3. Rumusan masalah.
4. Tujuan penelitian.
5. Data yang dibutuhkan.
6. Metode dan tahapan analisis.
7. Parameter yang digunakan beserta alasan pemilihannya.
8. Bentuk output/peta yang dihasilkan.
9. Interpretasi geofisika yang diharapkan.
10. Potensi pengembangan penelitian.

Gunakan bahasa Indonesia yang ilmiah tetapi tetap natural, jelas, dan mudah dipahami mahasiswa Geofisika. Hindari bahasa yang terlalu kaku dan jangan membuat penjelasan terlalu panjang.

**Q — Quality**  
Pastikan rancangan yang dibuat memiliki hubungan yang kuat antara permasalahan → data → metode → hasil → interpretasi geofisika. Topik harus realistis dikerjakan oleh mahasiswa, tidak membutuhkan terlalu banyak data yang sulit diperoleh, tetapi tetap memiliki novelty atau nilai penelitian yang cukup kuat.

Hindari topik yang terlalu umum seperti sekadar "pemetaan penggunaan lahan" tanpa keterkaitan dengan permasalahan geofisika. Utamakan topik yang menghasilkan zona prioritas, tingkat kerentanan, indeks spasial, atau pola distribusi suatu fenomena kebumian sehingga hasil SIG dapat dianalisis secara lebih mendalam.

## **2. RTCIKOQV — Total Station dalam Survei Geofisika**

**R — Role**  
Bertindaklah sebagai seorang ahli Geofisika dan Surveyor yang menguasai penggunaan Total Station untuk survei, pengukuran posisi, elevasi, pemetaan, dan akuisisi data geospasial dalam kegiatan geofisika.

**T — Task**  
Buatlah suatu rancangan penggunaan Total Station dalam kegiatan geofisika yang tidak hanya menjelaskan cara menggunakan alat, tetapi juga menunjukkan bagaimana data hasil pengukuran dapat dimanfaatkan untuk menghasilkan informasi geofisika yang akurat.

Fokuskan pembahasan pada pengukuran:
- koordinat titik $((X,Y))$,
- elevasi $((Z))$,
- jarak,
- sudut horizontal dan vertikal,
- serta perbedaan elevasi antar titik.

**C — Context**  
Total Station dapat digunakan sebagai alat pendukung dalam survei geofisika, terutama ketika diperlukan kontrol koordinat, pengukuran topografi, penentuan posisi titik pengukuran, pemetaan lintasan survei, dan koreksi geometrik data.

Kembangkan contoh kasus yang realistis, misalnya survei suatu area yang akan digunakan untuk akuisisi data geofisika. Beberapa titik pengukuran dibuat sebagai titik kontrol, kemudian posisi dan elevasinya diukur menggunakan Total Station.

Data hasil pengukuran dapat berupa:

| Titik | Easting (X) | Northing (Y) | Elevasi (Z) |
|---|---:|---:|---:|
| P1 | ... | ... | ... |
| P2 | ... | ... | ... |
| P3 | ... | ... | ... |

**I — Instruction**  
Jelaskan tahapan survei mulai dari persiapan alat, setup Total Station, centering dan leveling, backsight, pengukuran titik, pencatatan data, pengolahan koordinat, hingga visualisasi hasil survei.

Jelaskan pula prinsip dasar pengukuran Total Station dan hubungan antara sudut, jarak, serta koordinat titik. Jika diperlukan, gunakan konsep:

$$
\Delta X = D\sin\theta
$$

$$
\Delta Y = D\cos\theta
$$

serta:

$$
X_2 = X_1+\Delta X
$$

$$
Y_2 = Y_1+\Delta Y
$$

Untuk elevasi, jelaskan bagaimana pengukuran sudut vertikal atau zenith dan jarak dapat digunakan untuk memperoleh perbedaan tinggi antar titik.

Jangan hanya menjelaskan prosedur penggunaan alat. Hubungkan setiap tahapan dengan kebutuhan survei geofisika, terutama ketelitian posisi dan elevasi titik pengukuran.

**K — Knowledge**  
Gunakan konsep dasar:

- Total Station dan Electronic Distance Measurement (EDM).
- Horizontal angle dan vertical/zenith angle.
- Slope distance dan horizontal distance.
- Koordinat lokal dan koordinat terikat.
- Azimuth dan backsight.
- Benchmark (BM) dan titik kontrol.
- Traverse/poligon.
- Pengukuran beda tinggi.
- Error dan ketelitian pengukuran.
- Pemetaan topografi untuk mendukung survei geofisika.

Jelaskan bahwa kualitas koordinat titik pengukuran sangat penting karena posisi yang salah dapat memengaruhi interpretasi spasial data geofisika.

**O — Output**  
Hasilkan rancangan kegiatan dengan struktur:

1. Judul kegiatan/proyek.
2. Latar belakang.
3. Tujuan.
4. Konsep dasar Total Station.
5. Alat dan data yang diperlukan.
6. Tahapan akuisisi data.
7. Contoh tabel hasil pengukuran.
8. Perhitungan koordinat dan elevasi.
9. Analisis kesalahan/ketelitian.
10. Visualisasi titik dan lintasan survei.
11. Hubungan hasil pengukuran dengan kegiatan geofisika.
12. Interpretasi dan kesimpulan.

Berikan contoh hasil yang dapat diperoleh, seperti peta titik pengukuran, profil topografi, koordinat titik kontrol, dan model elevasi sederhana.

**Q — Quality**  
Pastikan penjelasan bersifat ilmiah, praktis, dan realistis seperti pekerjaan survei lapangan geofisika. Jangan menganggap Total Station sebagai metode geofisika utama, tetapi sebagai instrumen survei geospasial yang mendukung akuisisi dan interpretasi data geofisika.

Gunakan bahasa Indonesia yang natural, jelas, tidak terlalu kaku, dan mudah dipahami mahasiswa Geofisika. Jelaskan bagian matematis dengan benar dan jangan memasukkan konsep atau prosedur yang tidak relevan.

## **3. RTCIKOQV — Metode Geomagnetik (Magnetik)**

**R — Role**  
Bertindaklah sebagai seorang ahli Geofisika yang menguasai metode geomagnetik, khususnya akuisisi, pengolahan, pemodelan, dan interpretasi data medan magnet bumi untuk investigasi kondisi bawah permukaan.

**T — Task**  
Rancang suatu penelitian/proyek geofisika menggunakan metode geomagnetik yang menarik, realistis, dan memiliki nilai ilmiah. Fokuskan metode pada pemanfaatan variasi medan magnet untuk mengidentifikasi perbedaan sifat kemagnetan batuan dan struktur bawah permukaan**.

Proyek harus menunjukkan alur yang jelas mulai dari survei lapangan → akuisisi data magnetik → preprocessing → koreksi → pengolahan → pemetaan anomali → interpretasi geologi/geofisika.

**C — Context**  
Gunakan metode geomagnetik untuk menginvestigasi suatu wilayah yang memiliki kemungkinan variasi sifat magnetik bawah permukaan, misalnya keberadaan:

- kontak litologi,
- intrusi batuan beku,
- sesar atau struktur geologi,
- batuan vulkanik,
- zona mineralisasi,
- atau objek bawah permukaan lainnya.

Data utama dapat berupa intensitas medan magnet total (Total Magnetic Field/TMF) yang diukur menggunakan magnetometer. Jika tersedia, gunakan data base station untuk merekam variasi medan magnet harian.

Kembangkan contoh survei dengan beberapa titik pengukuran yang memiliki koordinat dan nilai medan magnet.

**I — Instruction**  
Jelaskan prinsip dasar metode geomagnetik dan mengapa batuan dengan suseptibilitas magnetik berbeda dapat menghasilkan anomali medan magnet.

Susun tahapan penelitian secara sistematis:

**Survei lapangan → Akuisisi TMI/TMF → Koreksi variasi harian → Koreksi IGRF → Data anomali magnetik → Filtering → Pemisahan regional-residual → Transformasi → Interpretasi**

Jika menggunakan koreksi IGRF, jelaskan konsep:

$$
\Delta T = T_{\text{observasi}} - T_{\text{IGRF}}
$$

Jelaskan bahwa pada data lapangan sebenarnya dapat terdapat pengaruh variasi medan magnet terhadap waktu sehingga data base station dapat digunakan untuk melakukan koreksi diurnal.

Gunakan beberapa metode pengolahan yang relevan, seperti:

- Reduction to Pole (RTP),
- upward continuation,
- downward continuation dengan pertimbangan kestabilan,
- first vertical derivative (FVD),
- horizontal derivative,
- regional-residual separation.

Jelaskan fungsi setiap metode dan kapan metode tersebut digunakan, bukan hanya menyebutkan namanya.

**K — Knowledge**  
Gunakan konsep dasar geomagnetik:

- medan magnet bumi,
- medan magnet utama,
- variasi diurnal,
- IGRF,
- Total Magnetic Intensity (TMI),
- anomali magnetik,
- suseptibilitas magnetik,
- magnetisasi induksi,
- magnetisasi remanen,
- dipole magnetic anomaly,
- regional dan residual anomaly.

Jelaskan hubungan sederhana:

$$
\mathbf{M} = \kappa \mathbf{H}
$$

dengan:

- $(\mathbf{M})$ = magnetisasi,
- $(\kappa)$ = suseptibilitas magnetik,
- $(\mathbf{H})$ = medan magnet yang menginduksi magnetisasi.

Interpretasi harus mempertimbangkan bahwa anomali magnetik tidak secara langsung menunjukkan jenis batuan tertentu. Interpretasi harus dikaitkan dengan informasi geologi, posisi anomali, bentuk anomali, pola gradien, dan jika memungkinkan data pendukung lainnya.

**O — Output**  
Hasilkan rancangan penelitian dengan struktur:

1. Judul penelitian.
2. Latar belakang.
3. Rumusan masalah.
4. Tujuan.
5. Konsep dasar metode geomagnetik.
6. Desain survei dan pola lintasan.
7. Data dan instrumen yang digunakan.
8. Tahapan akuisisi data.
9. Tahapan preprocessing dan koreksi.
10. Pengolahan data magnetik.
11. Peta anomali magnetik.
12. Interpretasi pola anomali.
13. Hubungan anomali dengan kondisi bawah permukaan.
14. Kesimpulan.

Jika memungkinkan, sertakan contoh bentuk peta TMI, peta anomali magnetik, peta RTP, peta derivative, dan penampang/anomali hasil interpretasi.

**Q — Quality**  
Pastikan penelitian tidak berhenti pada tahap "membuat peta anomali magnetik", tetapi menghasilkan interpretasi geofisika yang jelas mengenai kemungkinan struktur atau variasi sifat magnetik bawah permukaan.

Perhatikan kualitas data, spasi antar titik, arah lintasan, variasi diurnal, noise, posisi koordinat, dan resolusi survei. Jelaskan keterbatasan metode geomagnetik serta kemungkinan ambiguitas interpretasi.

Gunakan bahasa Indonesia yang ilmiah tetapi natural, jelas, dan mudah dipahami mahasiswa Geofisika. Buat rancangan yang realistis untuk penelitian lapangan dan memiliki potensi dikembangkan menjadi paper geofisika.

## **4. RTCIKOQV — Metode Geolistrik**

**R — Role**  
Bertindaklah sebagai seorang ahli Geofisika yang menguasai metode geolistrik, khususnya Electrical Resistivity Tomography (ERT), Vertical Electrical Sounding (VES), akuisisi data, pengolahan, inversi, dan interpretasi kondisi bawah permukaan berdasarkan distribusi resistivitas batuan.

**T — Task**  
Rancang suatu penelitian/proyek geofisika menggunakan metode geolistrik yang menarik, realistis, dan memiliki nilai ilmiah. Fokuskan penelitian pada pemanfaatan variasi resistivitas listrik bawah permukaan untuk mengidentifikasi kondisi geologi atau objek bawah permukaan.

Gunakan konfigurasi elektroda yang sesuai dengan tujuan survei, seperti Wenner, Wenner-Schlumberger, Dipole-Dipole, atau konfigurasi lainnya, kemudian jelaskan alasan pemilihannya.

**C — Context**  
Gunakan metode geolistrik untuk menginvestigasi permasalahan seperti:

- identifikasi lapisan bawah permukaan,
- zona air tanah,
- bidang kontak litologi,
- zona pelapukan,
- rekahan atau sesar,
- rongga/karst,
- zona mineralisasi,
- atau pencitraan struktur bawah permukaan.

Data lapangan dapat berupa posisi elektroda, konfigurasi elektroda, nilai arus $(I)$, beda potensial $(\Delta V)$, dan faktor geometrik $(K)$.

Hubungkan hasil pengukuran dengan sifat fisik material bawah permukaan, terutama **resistivitas listrik**.

**I — Instruction**  
Jelaskan prinsip dasar metode geolistrik berdasarkan hukum Ohm dan hubungan antara arus listrik, beda potensial, serta resistivitas.

Gunakan persamaan:

$$
\rho_a = K\frac{\Delta V}{I}
$$

dengan:

- $(\rho_a)$ = resistivitas semu,
- $(K)$ = faktor geometrik,
- $(\Delta V)$ = beda potensial,
- $(I)$ = arus listrik.

Jelaskan pula bahwa resistivitas semu belum tentu sama dengan resistivitas sebenarnya karena medan listrik dipengaruhi oleh kondisi bawah permukaan yang heterogen.

Susun tahapan penelitian:

**Desain survei → Penentuan lintasan → Akuisisi data → Perhitungan resistivitas semu → Quality control → Inversi → Model resistivitas 2D/3D → Interpretasi geofisika**

Jelaskan konfigurasi elektroda yang digunakan dan bagaimana posisi elektroda menentukan kedalaman investigasi, resolusi lateral, serta sensitivitas terhadap struktur tertentu.

Jika menggunakan ERT, jelaskan konsep inversi dan hubungan antara data observasi dengan model bawah permukaan:

$$
\mathbf{d} = \mathbf{Gm}
$$

serta konsep bahwa proses inversi digunakan untuk memperoleh model resistivitas yang paling sesuai dengan data pengukuran.

**K — Knowledge**  
Gunakan konsep dasar:

- resistivitas dan konduktivitas,
- hukum Ohm,
- arus listrik bawah permukaan,
- potensial listrik,
- faktor geometrik,
- resistivitas semu,
- konfigurasi elektroda,
- ERT,
- VES,
- forward modelling,
- inverse modelling,
- RMS error,
- sensitivity,
- regularisasi,
- dan interpretasi penampang resistivitas.

Interpretasi resistivitas tidak boleh dilakukan hanya berdasarkan angka. Pertimbangkan pula:

- litologi,
- kandungan air,
- porositas dan permeabilitas,
- tingkat pelapukan,
- salinitas,
- mineralisasi,
- struktur geologi,
- serta kondisi lingkungan setempat.

Hindari menyatakan bahwa satu nilai resistivitas pasti menunjukkan satu jenis batuan tertentu. Gunakan istilah **"diduga", "diinterpretasikan sebagai", atau "kemungkinan berkaitan dengan"** jika interpretasi memiliki ambiguitas.

**O — Output**  
Hasilkan rancangan penelitian dengan struktur:

1. Judul penelitian.
2. Latar belakang.
3. Rumusan masalah.
4. Tujuan penelitian.
5. Konsep dasar metode geolistrik.
6. Desain survei.
7. Konfigurasi elektroda.
8. Data dan instrumen.
9. Tahapan akuisisi.
10. Perhitungan resistivitas semu.
11. Quality control.
12. Pengolahan dan inversi.
13. Penampang resistivitas.
14. Interpretasi bawah permukaan.
15. Kesimpulan.

Tampilkan contoh bentuk output seperti pseudosection resistivitas semu, model resistivitas hasil inversi, penampang 2D, dan interpretasi zona bawah permukaan.

Jika memungkinkan, sertakan pula pembahasan mengenai kedalaman investigasi, resolusi, sensitivitas konfigurasi, dan RMS error hasil inversi.

**Q — Quality**  
Pastikan proyek tidak berhenti pada tahap menghasilkan penampang warna resistivitas, tetapi mampu menjelaskan hubungan antara pola resistivitas dengan kondisi bawah permukaan secara geofisika.

Perhatikan kualitas data, kontak elektroda, noise, spasi elektroda, panjang lintasan, konfigurasi yang digunakan, kedalaman investigasi, serta ketidakpastian hasil inversi.

Rancangan harus realistis untuk dilakukan oleh mahasiswa Geofisika dan memiliki potensi dikembangkan menjadi *penelitian lapangan atau paper geofisika.

Gunakan bahasa Indonesia yang ilmiah, natural, jelas, dan mudah dipahami, tanpa penjelasan yang terlalu kaku atau bertele-tele.

## **5. RTCIKOQV — Metode Penginderaan Jauh (Remote Sensing)**

**R — Role**  
Bertindaklah sebagai seorang ahli Geofisika dan Geoinformatika yang menguasai penginderaan jauh, citra satelit, pengolahan citra digital, analisis spasial, dan interpretasi fenomena kebumian berbasis data multispektral.

**T — Task**  
Rancang suatu penelitian/proyek penginderaan jauh untuk bidang geofisika dan geoinformatika yang menarik, realistis, dan memiliki nilai ilmiah. Penelitian harus memanfaatkan citra satelit untuk memperoleh informasi mengenai kondisi permukaan bumi dan menghubungkannya dengan suatu fenomena kebumian.

Fokus dapat diarahkan pada:
- perubahan tutupan atau penggunaan lahan,
- vegetasi,
- suhu permukaan,
- kondisi geologi,
- lingkungan,
- urbanisasi,
- kekeringan,
- atau fenomena kebumian lainnya.

**C — Context**  
Gunakan citra satelit yang relevan, seperti Sentinel-2, Landsat 8/9, Sentinel-1, atau sumber citra lainnya. Pemilihan citra harus disesuaikan dengan fenomena yang ingin dikaji.

Sebagai contoh, penelitian dapat menggunakan citra multispektral untuk memperoleh indeks seperti **NDVI, NDWI, NDBI, atau LST**, kemudian mengintegrasikannya dengan data spasial lainnya untuk menghasilkan informasi geospasial yang lebih bermakna.

Pastikan penelitian memiliki wilayah studi yang jelas dan mempertimbangkan **resolusi spasial, temporal, dan spektral** dari data yang digunakan.

**I — Instruction**  
Susun penelitian mulai dari **pemilihan citra → preprocessing → koreksi/penyaringan data → ekstraksi informasi → analisis spasial → klasifikasi atau indeks → validasi → interpretasi**.

Alur penelitian dapat dibuat seperti:

**Citra Satelit → Preprocessing → Ekstraksi Parameter → Analisis Spasial → Integrasi Data → Peta Tematik → Validasi → Interpretasi**

Jika menggunakan indeks vegetasi, jelaskan dasar matematisnya, misalnya:

$$
NDVI = \frac{NIR-Red}{NIR+Red}
$$

Jika menggunakan NDWI:

$$
NDWI = \frac{Green-NIR}{Green+NIR}
$$

Jika menggunakan NDBI:

$$
NDBI = \frac{SWIR-NIR}{SWIR+NIR}
$$

Jelaskan fungsi masing-masing indeks dan alasan pemilihannya terhadap permasalahan yang dikaji.

**K — Knowledge**  
Gunakan konsep dasar penginderaan jauh, seperti:

- resolusi spasial,
- resolusi spektral,
- resolusi temporal,
- resolusi radiometrik,
- spektrum elektromagnetik,
- reflektansi,
- band multispektral,
- preprocessing citra,
- cloud masking,
- compositing,
- klasifikasi citra,
- indeks spektral,
- supervised dan unsupervised classification,
- serta validasi hasil klasifikasi.

Jelaskan hubungan antara karakteristik spektral objek dengan informasi yang diperoleh dari citra. Jangan hanya menyebutkan band, tetapi jelaskan mengapa band tertentu digunakan untuk mengekstraksi suatu fenomena.

Jika menggunakan klasifikasi, gunakan metrik validasi yang sesuai, seperti:

$$
Overall\ Accuracy = \frac{\text{Jumlah piksel yang diklasifikasikan dengan benar}}
{\text{Total piksel validasi}}
\times 100\%
$$

Jika memungkinkan, gunakan confusion matrix, overall accuracy, producer's accuracy, user's accuracy, dan kappa coefficient untuk mengevaluasi hasil klasifikasi.

**O — Output**  
Hasilkan rancangan penelitian dengan struktur:

1. Judul penelitian.
2. Latar belakang.
3. Rumusan masalah.
4. Tujuan penelitian.
5. Data yang digunakan.
6. Karakteristik citra.
7. Tahapan preprocessing.
8. Metode ekstraksi informasi.
9. Analisis spasial.
10. Validasi.
11. Peta hasil.
12. Interpretasi geofisika/geoinformatika.
13. Kesimpulan.

Output utama dapat berupa citra hasil preprocessing, peta indeks spektral, peta klasifikasi, peta perubahan, atau peta distribusi fenomena tertentu.

Hubungkan hasil penginderaan jauh dengan kondisi nyata di lapangan dan jelaskan bagaimana informasi tersebut dapat dimanfaatkan untuk pemetaan, monitoring, mitigasi, atau penelitian kebumian.

**Q — Quality**  
Pastikan penelitian tidak berhenti pada tahap menghasilkan citra atau peta indeks, tetapi memiliki analisis yang menjawab suatu permasalahan geoinformatika/geofisika.

Perhatikan kualitas citra, tutupan awan, resolusi data, pemilihan band, preprocessing, metode klasifikasi, validasi, serta keterbatasan interpretasi citra.

Jika hasil digunakan untuk menginterpretasikan kondisi bawah permukaan atau fenomena geofisika, jelaskan bahwa citra satelit terutama memberikan informasi permukaan, sehingga interpretasi bawah permukaan harus didukung oleh data geologi, geofisika, topografi, atau data lapangan lainnya.

Buat rancangan yang realistis untuk mahasiswa Geoinformatika/Geofisika, memiliki nilai kebaruan, dapat dikerjakan dengan data terbuka, dan berpotensi dikembangkan menjadi penelitian atau paper ilmiah.

Gunakan bahasa Indonesia yang ilmiah tetapi natural, jelas, dan tidak terlalu kaku atau bertele-tele.
