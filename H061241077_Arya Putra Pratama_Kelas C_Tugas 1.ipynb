## **3. Implementasi Python**
### **3.1 Mengimpor Pustaka**
Pustaka yang digunakan dibatasi pada kebutuhan utama pengolahan data dan visualisasi.

```python
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### **3.2 Mengambil Data Gempa dari USGS**

Data diambil menggunakan USGS Earthquake Catalog API dalam format GeoJSON. Jika pengambilan data gagal, program akan menampilkan pesan kesalahan.

```python
url = "https://earthquake.usgs.gov/fdsnws/event/1/query"

params = {
    "format": "geojson",
    "starttime": "2016-01-01",
    "endtime": "2026-10-13",
    "minlatitude": -2.5,
    "maxlatitude": 0.0,
    "minlongitude": 119.0,
    "maxlongitude": 121.0,
    "minmagnitude": 2.5,
    "orderby": "time-asc",
    "limit": 20000}

try:
    response = requests.get(url, params=params, timeout=60)
    
    response.raise_for_status()
    hasil = response.json()

    fitur = hasil.get("features", [])

    if not fitur:
        raise ValueError(
            "Tidak ditemukan data gempa dengan kriteria tersebut.")

    print("Jumlah kejadian yang diperoleh:", len(fitur))

except (requests.RequestException, ValueError) as error:
    raise RuntimeError(
        f"Pengambilan data gagal: {error}")
```

Jumlah kejadian yang diperoleh bergantung pada hasil aktual katalog USGS saat kode dijalankan. Batas API yang digunakan adalah 20.000 kejadian; apabila batas tersebut tercapai, pengambilan data perlu diperiksa agar tidak menganggap katalog sudah lengkap.

### **3.3 Menyusun dan Membersihkan Data**
Informasi penting diekstrak dari GeoJSON, kemudian disusun menjadi tabel menggunakan pandas.

```python
data = []

for gempa in fitur:
    properti = gempa["properties"]
    koordinat = gempa["geometry"]["coordinates"]

    data.append({
        "time": properti.get("time"),
        "longitude": koordinat[0],
        "latitude": koordinat[1],
        "depth": koordinat[2],
        "magnitude": properti.get("mag"),
        "place": properti.get("place"),
        "id": gempa.get("id")})

df = pd.DataFrame(data)

df["time"] = pd.to_datetime(df["time"], unit="ms", utc=True)

kolom_numerik = ["longitude", "latitude", "depth", "magnitude"]

df[kolom_numerik] = df[kolom_numerik].apply(
    pd.to_numeric,
    errors="coerce")

jumlah_awal = len(df)

df = df.dropna(
    subset=["time", "longitude", "latitude", "depth", "magnitude"])

df = df[
    df["longitude"].between(119.0, 121.0)
    & df["latitude"].between(-2.5, 0.0)
    & (df["depth"] >= 0)
    & (df["magnitude"] >= 2.5)]

df = df.drop_duplicates(subset="id")
df = df.sort_values("time").reset_index(drop=True)

print("Jumlah data awal:", jumlah_awal)
print("Jumlah data setelah pemeriksaan:", len(df))
print("Jumlah data yang dihapus:", jumlah_awal - len(df))

display(df.head())
```

Pemeriksaan tersebut menghilangkan data dengan atribut penting yang kosong, koordinat di luar batas penelitian, kedalaman negatif, magnitudo di bawah ambang, dan ID kejadian yang berulang.

### **3.4 Statistik Deskriptif**
Statistik deskriptif digunakan untuk melihat karakteristik umum katalog gempa.

```python
if df.empty:
    raise ValueError("Tidak ada data valid untuk dianalisis.")

statistik = pd.Series({
    "Jumlah gempa": len(df),
    "Magnitudo rata-rata": df["magnitude"].mean(),
    "Magnitudo maksimum": df["magnitude"].max(),
    "Kedalaman rata-rata (km)": df["depth"].mean(),
    "Kedalaman maksimum (km)": df["depth"].max()})

display(statistik.round(2).to_frame("Nilai"))

print("\nStatistik lengkap:")
display (df[["magnitude", "depth"]].describe().round(2))
```

Hasil statistik menunjukkan nilai magnitudo dan kedalaman yang tercatat dalam katalog. Magnitudo rata-rata tidak menggambarkan energi total seluruh gempa karena skala magnitudo bersifat logaritmik.

### **3.5 Distribusi Gempa Berdasarkan Waktu**
Jumlah kejadian dikelompokkan berdasarkan tahun untuk melihat perubahan jumlah gempa selama periode pengamatan.

```python
gempa_tahunan = (df.groupby(df["time"].dt.year).size())

display(gempa_tahunan.rename("Jumlah gempa").to_frame())
```

Pengelompokan ini hanya digunakan sebagai statistik pendukung. Perubahan jumlah kejadian dapat dipengaruhi oleh aktivitas seismik, kelengkapan katalog, dan kemampuan deteksi jaringan seismik.

## **4. Visualisasi Utama**
### **4.1 Peta Episenter Gempa**
Peta ini menunjukkan posisi episenter berdasarkan longitude dan latitude. Warna mewakili kedalaman hiposenter, sedangkan ukuran simbol menunjukkan magnitudo.

```python
fig, ax = plt.subplots(figsize=(10, 8), dpi=150)

ukuran = 15 + 12 * (
    df["magnitude"] - df["magnitude"].min())

scatter = ax.scatter(
    df["longitude"],
    df["latitude"],
    c=df["depth"],
    s=ukuran,
    cmap="viridis_r",
    alpha=0.75,
    edgecolors="black",
    linewidths=0.25)

cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label("Kedalaman hiposenter (km)")

ax.set_xlim(119.0, 121.0)
ax.set_ylim(-2.5, 0.0)

ax.set_xlabel("Longitude (° BT)")
ax.set_ylabel("Latitude (° LS)")
ax.set_title(
    "Peta Episenter Gempa di Sekitar Sesar Palu-Koro\n"
    "USGS Earthquake Catalog, 2016–2026")

ax.grid(
    True,
    linestyle="--",
    alpha=0.3)

ax.text(
    0.02, 0.02,
    "Ukuran simbol menunjukkan magnitudo\n"
    "Sumber: USGS Earthquake Catalog",
    transform=ax.transAxes,
    fontsize=8,
    verticalalignment="bottom")

plt.tight_layout()
plt.show()
```

### **4.2 Penampang Hiposenter**

Penampang dibuat mengikuti lintasan berarah barat laut–tenggara yang didefinisikan oleh dua titik geografis. Posisi gempa diproyeksikan ke arah lintasan menggunakan koordinat lokal dalam kilometer. Hanya gempa dalam koridor 25 km dari lintasan yang ditampilkan.

```python
lat0, lon0 = -0.3, 119.7
lat1, lon1 = -2.2, 120.1

lat_ref = np.radians((lat0 + lat1) / 2)

def koordinat_lokal(lon, lat):
    x = ((np.asarray(lon) - lon0)* 111.32* np.cos(lat_ref))
    y = (np.asarray(lat) - lat0) * 111.32
    return x, y

x0, y0 = 0.0, 0.0

x1, y1 = koordinat_lokal(lon1, lat1)

panjang = np.hypot(x1, y1)

if panjang == 0:
    raise ValueError("Titik awal dan akhir lintasan harus berbeda.")

ux = x1 / panjang
uy = y1 / panjang

x, y = koordinat_lokal(df["longitude"],df["latitude"])

df["jarak_lintasan"] = x * ux + y * uy
df["jarak_silang"] = np.abs(x * uy - y * ux)

penampang = df[
    (df["jarak_lintasan"] >= 0)
    & (df["jarak_lintasan"] <= panjang)
    & (df["jarak_silang"] <= 25)].copy()

print("Panjang lintasan (km):", round(panjang, 2))
print("Jumlah gempa pada penampang:", len(penampang))

if penampang.empty:
    raise ValueError(
        "Tidak ada gempa dalam koridor penampang. "
        "Periksa kembali lintasan atau lebar koridor.")

fig, ax = plt.subplots(figsize=(10, 6), dpi=150)

ukuran = 15 + 12 * (penampang["magnitude"] - penampang["magnitude"].min())

scatter = ax.scatter(
    penampang["jarak_lintasan"],
    penampang["depth"],
    s=ukuran,
    c=penampang["magnitude"],
    cmap="plasma",
    alpha=0.8,
    edgecolors="black",
    linewidths=0.3)

cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label("Magnitudo")

ax.set_xlabel("Jarak sepanjang lintasan (km)")
ax.set_ylabel("Kedalaman hiposenter (km)")
ax.set_title(
    "Penampang Hiposenter Gempa di Sekitar Sesar Palu-Koro")

ax.invert_yaxis()
ax.grid(True, linestyle="--", alpha=0.3)

ax.text(
    0.02, 0.02,
    "Ukuran simbol menunjukkan magnitudo\n"
    "Koridor penampang: ±25 km",
    transform=ax.transAxes,
    fontsize=8,
    verticalalignment="bottom")

plt.tight_layout()
plt.show()
```

**Interpretasi penampang:**

Penampang menunjukkan perubahan kedalaman hiposenter terhadap jarak sepanjang lintasan. Perhatikan apakah gempa terkonsentrasi pada kedalaman tertentu atau tersebar pada rentang kedalaman yang lebih luas.

Penampang ini merupakan proyeksi geografis sederhana, bukan penampang struktur sesar yang sebenarnya. Gempa yang jauh dari lintasan tidak ditampilkan, sedangkan kedalaman yang tercatat tetap memiliki ketidakpastian. ## **2. Landasan Teori**
### **2.1 Anomali Magnetik**
Anomali magnetik merupakan variasi medan magnet yang berkaitan dengan keberadaan sumber magnetik di bawah permukaan. Perbedaan sifat kemagnetan batuan dapat menghasilkan respons yang berbeda pada titik pengamatan.

Salah satu pendekatan sederhana untuk memodelkan sumber magnetik adalah dipol magnetik. Model ini menggambarkan sumber sebagai momen magnetik yang terpusat pada satu titik.

### **2.2 Persamaan Medan Magnet Dipol**
Medan magnet yang dihasilkan oleh dipol ideal dinyatakan sebagai:

$$
\mathbf{B}(\mathbf{r}) =
\frac{\mu_0}{4\pi r^3}
\left[
3(\mathbf{m}\cdot\hat{\mathbf{r}})
\hat{\mathbf{r}}-\mathbf{m}
\right]
$$

Keterangan:

* $\mathbf{B}$ = medan magnet dipol (T).
* $\mu_0$ = permeabilitas magnetik vakum, yaitu $4\pi\times10^{-7}$ H/m.
* $\mathbf{m}$ = momen dipol magnetik (A m²).
* $r$ = jarak antara sumber dan titik pengamatan (m).
* $\hat{\mathbf{r}}$ = vektor satuan dari sumber menuju titik pengamatan.

Persamaan tersebut menunjukkan bahwa medan magnet dipol dipengaruhi oleh besar momen magnetik, jarak pengamatan, dan orientasi sumber. Untuk orientasi tetap, besar medan dipol berkurang sebanding dengan $1/r^3$.

### **2.3 Anomali Medan Magnet Total**
Dalam pendekatan medan anomali yang relatif kecil dibandingkan medan magnet utama, anomali medan magnet total dapat dihampiri dengan proyeksi medan anomali pada arah medan utama:

$$
\Delta T \approx \hat{\mathbf{b}}_0\cdot\mathbf{B}
$$

dengan:

* $\Delta T$ = anomali medan magnet total (T).
* $\hat{\mathbf{b}}_0$ = vektor satuan arah medan magnet utama.
* $\mathbf{B}$ = medan anomali yang dihasilkan oleh sumber.

Dalam simulasi ini, arah medan utama dan arah momen dipol ditentukan sebagai parameter asumsi. Oleh karena itu, hasilnya merupakan demonstrasi numerik, bukan representasi kondisi magnetik aktual suatu wilayah.

Konversi satuan yang digunakan adalah:

$$
1\ \text{T}=10^9\ \text{nT}
$$

## **4. Implementasi Python**
### **4.1 Mengimpor Pustaka**
Gunakan NumPy untuk perhitungan numerik dan Matplotlib untuk visualisasi. Tidak diperlukan data eksternal maupun file tambahan.

```python
import numpy as np
import matplotlib.pyplot as plt

MU0_4PI = 1e-7  # μ0 / 4π dalam SI
```

### **4.2 Menentukan Parameter dan Fungsi Pemodelan**
Sistem koordinat yang digunakan adalah X ke timur, Y ke utara, dan Z ke atas. Sumber berada pada koordinat $(0,0,-h)$, sedangkan titik pengamatan berada pada permukaan $(x,y,0)$.

Momen dipol dibuat searah dengan medan magnet utama asumsi. Fungsi berikut menghitung tiga komponen medan dipol dan proyeksinya pada arah medan utama.

```python
D = np.deg2rad(0.0)
I = np.deg2rad(45.0)

arah_medan = np.array([
    np.cos(I) * np.sin(D),
    np.cos(I) * np.cos(D),
    -np.sin(I)
])

def model_dipol(X, Y, kedalaman, momen):
    """
    Menghitung anomali medan magnet total pendekatan dipol.

    X, Y    : koordinat pengamatan (m)
    kedalaman : kedalaman sumber (m)
    momen   : besar momen dipol (A m²)

    Return:
    Anomali medan magnet total pendekatan dalam nT.
    """

    rx = X
    ry = Y
    rz = kedalaman

    r2 = rx**2 + ry**2 + rz**2
    r = np.sqrt(r2)

    mx, my, mz = momen * arah_medan

    dot_mr = mx * rx + my * ry + mz * rz

    faktor = MU0_4PI / r**5

    Bx = faktor * (3 * dot_mr * rx - mx * r2)
    By = faktor * (3 * dot_mr * ry - my * r2)
    Bz = faktor * (3 * dot_mr * rz - mz * r2)

    delta_T = (
        arah_medan[0] * Bx
        + arah_medan[1] * By
        + arah_medan[2] * Bz)

    return delta_T * 1e9  # Konversi T ke nT
```

Fungsi tersebut menghitung pendekatan anomali medan magnet total berdasarkan proyeksi medan dipol. Fungsi tidak menghitung medan magnet utama bumi itu sendiri.

### **4.3 Membuat Grid Pengamatan**
Grid dua dimensi dibuat untuk menghitung respons magnetik pada beberapa titik di permukaan.

```python
x = np.linspace(-2000, 2000, 101)
y = np.linspace(-2000, 2000, 101)

X, Y = np.meshgrid(x, y)

kedalaman_awal = 500.0
momen_awal = 1000.0

anomali = model_dipol(
    X, Y,
    kedalaman_awal,
    momen_awal)

print("Ukuran grid:", anomali.shape)
print("Anomali minimum (nT):", round(anomali.min(), 3))
print("Anomali maksimum (nT):", round(anomali.max(), 3))
```

## **5. Hasil dan Visualisasi**
### **5.1 Peta Kontur Anomali Magnetik 2D**
Peta kontur menunjukkan distribusi anomali magnetik pada permukaan di sekitar sumber. Warna menunjukkan nilai anomali dalam nanotesla.

```python
fig, ax = plt.subplots(figsize=(9, 7), dpi=150)

kontur = ax.contourf(X, Y, anomali,levels=30,cmap="RdBu_r")
ax.contour(X, Y, anomali,levels=10,colors="black",linewidths=0.45,alpha=0.6)

cbar = plt.colorbar(kontur, ax=ax)
cbar.set_label("Anomali medan magnet total (nT)")

ax.scatter(
    0, 0,
    marker="*",
    s=150,
    color="black",
    label="Proyeksi posisi sumber")

ax.set_xlabel("X (m)")
ax.set_ylabel("Y (m)")
ax.set_title(
    "Peta Kontur Anomali Magnetik 2D\n"
    "Model Dipol Sintetis")

ax.set_aspect("equal")
ax.legend()
plt.tight_layout()
plt.show()
```

**Interpretasi:**

Peta memperlihatkan perubahan nilai anomali di sekitar proyeksi posisi sumber. Pola positif dan negatif muncul akibat orientasi momen dipol serta arah pengamatan terhadap sumber. Bentuk anomali tidak harus simetris secara sederhana karena medan dipol merupakan medan vektor. Bintang menandai proyeksi horizontal posisi sumber, bukan lokasi pengukuran ataupun batas tubuh geologi.

### **5.2 Profil Anomali Magnetik terhadap Kedalaman**
Profil dibuat dengan membandingkan tiga kedalaman sumber, sementara momen dipol, orientasi, dan posisi pengamatan tetap sama.


```python
x_profil = np.linspace(-2000, 2000, 401)
y_profil = np.zeros_like(x_profil)

kedalaman_uji = [500, 1000, 1500]

fig, ax = plt.subplots(figsize=(9, 5.5), dpi=150)
for h in kedalaman_uji:
    profil = model_dipol(x_profil,y_profil,h,momen_awal)

    ax.plot(
        x_profil,
        profil,
        linewidth=2,
        label=f"Kedalaman {h} m")

ax.axhline(0,color="black",linewidth=0.8,alpha=0.6)

ax.set_xlabel("Jarak sepanjang profil X (m)")
ax.set_ylabel("Anomali medan magnet total (nT)")
ax.set_title("Perbandingan Profil Anomali pada Kedalaman Berbeda")
ax.grid(True, linestyle="--", alpha=0.3)
ax.legend()

plt.tight_layout()
plt.show()
```

**Interpretasi:**

Bandingkan amplitudo dan lebar anomali untuk setiap kedalaman. Sumber yang lebih dalam umumnya menghasilkan anomali dengan amplitudo lebih rendah dan pola yang lebih lebar. Kesimpulan tersebut berlaku pada perbandingan dengan momen dipol, orientasi, dan konfigurasi pengamatan yang sama.

### **5.3 Menguji Pengaruh Kekuatan Momen Dipol**
Untuk mengamati pengaruh kekuatan sumber, kedalaman dipertahankan tetap dan momen dipol divariasikan.

```python
momen_uji = [500, 1000, 2000]
kedalaman_tetap = 500.0

fig, ax = plt.subplots(figsize=(9, 5.5), dpi=150)

for m in momen_uji:
    profil = model_dipol(x_profil,y_profil,kedalaman_tetap,m)
    ax.plot(x_profil,profil,linewidth=2,label=f"Momen {m} A m²")

ax.axhline(
    0,
    color="black",
    linewidth=0.8,
    alpha=0.6)

ax.set_xlabel("Jarak sepanjang profil X (m)")
ax.set_ylabel("Anomali medan magnet total (nT)")
ax.set_title("Pengaruh Momen Dipol terhadap Anomali Magnetik")
ax.grid(True, linestyle="--", alpha=0.3)
ax.legend()

plt.tight_layout()
plt.show()
```

**Interpretasi:**

Dengan kedalaman dan orientasi yang tetap, amplitudo anomali berbanding lurus dengan besar momen dipol pada model linear ini. Menggandakan momen dipol akan menggandakan nilai anomali pada setiap titik pengamatan, tanpa mengubah bentuk relatif profil.

Metode *Electrical Resistivity Tomography* (ERT) merupakan metode geofisika yang digunakan untuk mempelajari variasi resistivitas listrik bawah permukaan. Perbedaan resistivitas dapat berkaitan dengan variasi jenis batuan, kandungan air, porositas, dan kondisi material bawah permukaan.

Dalam penelitian ini, Python digunakan untuk membuat model resistivitas sintetis dua dimensi yang terdiri atas beberapa lapisan dan satu anomali lokal. Model tersebut digunakan untuk memahami bagaimana perbedaan resistivitas dapat divisualisasikan dan diinterpretasikan secara geofisika.

### **1.2 Tujuan Penelitian**

2. Mengamati perbedaan resistivitas antarlapisan dan anomali lokal.

## **2. Dasar Teori**
### **2.1 Resistivitas Batuan**
Resistivitas merupakan ukuran kemampuan material dalam menghambat aliran arus listrik. Satuannya adalah ohm-meter ($(\Omega\,\text{m})$).

Secara umum, material dengan resistivitas rendah lebih mudah menghantarkan arus listrik dibandingkan material dengan resistivitas tinggi. Namun, nilai resistivitas juga dipengaruhi oleh porositas, kandungan fluida, salinitas, temperatur, dan komposisi mineral.

### **2.2 Prinsip Dasar ERT**

Pada survei ERT, arus listrik dialirkan melalui elektroda arus, sedangkan beda potensial diukur menggunakan elektroda potensial. Resistivitas semu dihitung menggunakan persamaan:

$$
\rho_a=K\frac{\Delta V}{I}
$$

Keterangan:

* $(\rho_a)$ = resistivitas semu ($(\Omega\,\text{m})$).
* $(K)$ = faktor geometri elektroda (m).
* $(\Delta V)$ = beda potensial (V).
* $(I)$ = arus listrik (A).

Resistivitas semu merupakan respons gabungan material yang dipengaruhi konfigurasi elektroda dan distribusi resistivitas bawah permukaan. Oleh karena itu, resistivitas semu tidak sama dengan resistivitas sebenarnya pada satu titik atau kedalaman tertentu.

Dalam proyek ini, model resistivitas sebenarnya dibuat secara sintetis. Pemodelan respons elektroda dan inversi tidak dilakukan, sehingga hasilnya digunakan untuk mempelajari struktur model, bukan sebagai hasil survei ERT.

## **3. Parameter dan Model Sintetis**

Model bawah permukaan terdiri atas tiga lapisan dan satu anomali lokal.

Nilai-nilai tersebut merupakan asumsi simulasi, bukan hasil pengukuran lapangan. Anomali dengan resistivitas rendah digunakan sebagai contoh zona yang lebih konduktif daripada material di sekitarnya.

## **4. Implementasi Python**
### **4.1 Mengimpor Pustaka**
Pustaka yang digunakan hanya NumPy untuk membuat model dan Matplotlib untuk visualisasi.

```python
import numpy as np
import matplotlib.pyplot as plt
```

### **4.2 Menentukan Parameter Model**

```python
panjang = 200  
kedalaman = 60 

dx = 2
dz = 1

x = np.arange(0, panjang + dx, dx)
z = np.arange(0, kedalaman + dz, dz)

X, Z = np.meshgrid(x, z)

rho1 = 50
rho2 = 150
rho3 = 500

rho = np.zeros_like(X, dtype=float)

rho[Z < 10] = rho1
rho[(Z >= 10) & (Z < 30)] = rho2
rho[Z >= 30] = rho3

x_anomali = 110
z_anomali = 22

lebar_anomali = 30
tinggi_anomali = 12

rho_anomali = 20
```

### **4.3 Menambahkan Anomali Lokal**
Anomali dibuat berbentuk persegi panjang untuk merepresentasikan zona yang memiliki resistivitas berbeda dari lingkungan sekitarnya.

```python
mask_anomali = (
    (np.abs(X - x_anomali) <= lebar_anomali / 2)
    & (np.abs(Z - z_anomali) <= tinggi_anomali / 2))

rho[mask_anomali] = rho_anomali

print("Ukuran model:", rho.shape)
print("Resistivitas minimum:", rho.min(), "ohm-meter")
print("Resistivitas maksimum:", rho.max(), "ohm-meter")
```

Model yang dihasilkan merupakan distribusi resistivitas sebenarnya yang ditentukan langsung dari parameter sintetis. Belum ada simulasi injeksi arus, pengukuran beda potensial, atau inversi.

## **5. Visualisasi Model Resistivitas**
### **5.1 Model Resistivitas Sebenarnya (*True Resistivity Model*)**
Penampang berikut memperlihatkan susunan lapisan dan anomali lokal berdasarkan nilai resistivitas yang telah ditentukan.

```python
fig, ax = plt.subplots(figsize=(10, 5.5), dpi=150)

kontur = ax.pcolormesh(
    X,
    Z,
    rho,
    shading="auto",
    cmap="turbo",
    vmin=20,
    vmax=500)

cbar = plt.colorbar(kontur, ax=ax)
cbar.set_label("Resistivitas (Ω m)")

ax.set_xlabel("Jarak horizontal (m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("Model Resistivitas Bawah Permukaan 2D")

ax.invert_yaxis()

ax.set_xlim(0, panjang)
ax.set_ylim(kedalaman, 0)

ax.set_aspect("auto")
ax.grid(False)

plt.tight_layout()
plt.show()
```

**Interpretasi geofisika:**

Penampang menunjukkan tiga lapisan dengan resistivitas berbeda. Lapisan pertama memiliki resistivitas $(50\ \Omega\,\text{m})$, lapisan kedua $(150\ \Omega\,\text{m})$, dan lapisan ketiga $(500\ \Omega\,\text{m})$. Anomali lokal memiliki resistivitas $(20\ \Omega\,\text{m})$, sehingga lebih konduktif daripada material di sekitarnya.

Secara hipotetis, zona konduktif dapat berkaitan dengan material lempung atau material yang memiliki kandungan fluida lebih konduktif. Akan tetapi, interpretasi tersebut belum dapat dipastikan hanya berdasarkan nilai resistivitas sintetis.

### 5.2 Profil Resistivitas terhadap Kedalaman
Sebagai visualisasi pendukung, profil vertikal diambil pada posisi horizontal yang melewati pusat anomali. Profil ini menunjukkan perubahan resistivitas model terhadap kedalaman, bukan pseudosection hasil pengukuran.

```python

indeks_x = np.argmin(np.abs(x - x_anomali))

profil_rho = rho[:, indeks_x]

fig, ax = plt.subplots(figsize=(6, 6), dpi=150)

ax.step(profil_rho,z,where="mid",linewidth=2)

ax.set_xlabel("Resistivitas (Ω m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("Profil Resistivitas pada Pusat Anomali")

ax.invert_yaxis()
ax.grid(True, linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

**Interpretasi geofisika:**

Profil memperlihatkan perubahan resistivitas sesuai batas lapisan dan keberadaan anomali lokal. Pada kedalaman yang berpotongan dengan anomali, resistivitas turun menjadi $(20\ \Omega\,\text{m})$, kemudian kembali mengikuti nilai lapisan di sekitarnya ketika profil keluar dari anomali.

Profil ini membantu mengamati perubahan vertikal pada model yang telah dibuat. Profil tersebut bukan hasil inversi geolistrik dan tidak menunjukkan resolusi deteksi elektroda.

## **2. Dasar Teori**

### **2.1 Prinsip Dasar InSAR**

SAR merupakan sistem radar aktif yang merekam respons hamburan balik permukaan bumi. Setiap piksel citra SAR memiliki informasi amplitudo dan fase. Dengan membandingkan fase dari dua citra yang direkam pada waktu berbeda, perubahan jarak sepanjang garis pandang sensor dapat dianalisis.

Perbedaan fase tersebut membentuk interferogram. Namun, fase interferometrik tidak hanya dipengaruhi deformasi, melainkan juga geometri orbit, topografi, atmosfer, dan derau. Dalam simulasi sederhana ini, kontribusi utama yang dimodelkan adalah fase akibat deformasi.

### **2.2 Model Deformasi Mogi**

Model Mogi merupakan pendekatan elastik sederhana untuk menggambarkan deformasi permukaan akibat perubahan volume sumber tekanan yang berada di bawah permukaan bumi. Model ini mengasumsikan medium homogen, isotropik, elastik, dan sumber yang relatif kecil dibandingkan kedalamannya.

Besarnya deformasi vertikal dapat dinyatakan sebagai:

$$
u_z(r)=\frac{(1-\nu)\Delta V}{\pi}
\frac{d}{(r^2+d^2)^{3/2}}
$$

Sementara itu, besar perpindahan horizontal radial adalah:

$$
u_r(r)=\frac{(1-\nu)\Delta V}{\pi}
\frac{r}{(r^2+d^2)^{3/2}}
$$

dengan:

* $(u_z)$ = perpindahan vertikal permukaan (m).
* $(u_r)$ = perpindahan horizontal radial (m).
* $(r)$ = jarak horizontal dari pusat sumber (m).
* $(d)$ = kedalaman sumber (m).
* $(\Delta V)$ = perubahan volume sumber (m³).
* $(\nu)$ = rasio Poisson batuan.

Dalam model ini, $(\Delta V>0)$ menghasilkan pengangkatan permukaan, sedangkan $(\Delta V<0)$ menghasilkan penurunan permukaan.

### **2.3 Deformasi LOS dan Fase InSAR**

InSAR mengukur perubahan jarak dalam arah *Line of Sight* (LOS), yaitu arah antara sensor radar dan permukaan. InSAR tidak secara langsung mengukur seluruh komponen perpindahan tiga dimensi.

Untuk geometri pengamatan dengan sudut datang $(\theta)$, serta arah pandang horizontal dari satelit menuju permukaan dengan azimut $(\alpha)$, deformasi LOS dapat dihitung melalui proyeksi vektor perpindahan:

$$
d_{\mathrm{LOS}} =
-u_E\sin\theta\sin\alpha
-u_N\sin\theta\cos\alpha
+u_z\cos\theta
$$

Pada persamaan tersebut, $(u_E)$, $(u_N)$, dan $(u_z)$ masing-masing merupakan perpindahan ke timur, utara, dan vertikal. Konvensi tanda yang digunakan adalah **positif untuk perpindahan menuju sensor**.

Hubungan antara deformasi LOS dan perubahan fase akibat deformasi adalah:

$$
\Delta\phi_{\mathrm{def}}=
-\frac{4\pi}{\lambda}d_{\mathrm{LOS}}
$$

dengan $(\lambda)$ merupakan panjang gelombang radar dalam meter. Pada simulasi ini, fase terbungkus dihitung menggunakan:

$$
\phi_{\mathrm{wrapped}}=
\operatorname{atan2}
\left(\sin\Delta\phi_{\mathrm{def}},
\cos\Delta\phi_{\mathrm{def}}\right)
$$

Fase terbungkus berada dalam rentang $([-\pi,\pi])$. Oleh karena itu, satu nilai fase dapat mewakili beberapa kemungkinan perubahan jarak yang berbeda sebesar kelipatan panjang gelombang setengah. Pada data nyata, diperlukan tahapan pengolahan tambahan, termasuk koreksi komponen fase lain dan *phase unwrapping*, sebelum deformasi LOS dapat ditafsirkan dengan lebih lengkap.

## 3. Parameter dan Model Sintetis

Simulasi menggunakan area berbentuk persegi dengan sumber deformasi berada di bawah pusat area.

| Parameter                     |           Nilai | Satuan  | Keterangan                                                  |
| ----------------------------- | --------------: | ------- | ----------------------------------------------------------- |
| Panjang gelombang radar       |           0,056 | m       | Pendekatan panjang gelombang Sentinel-1 pita C              |
| Luas area pengamatan          |         10 × 10 | km      | Area simulasi sintetis                                      |
| Resolusi grid                 |             0,1 | km      | Jarak antartitik grid                                       |
| Posisi sumber $(x_0)$         |               0 | km      | Pusat sumber arah timur                                     |
| Posisi sumber $(y_0)$         |               0 | km      | Pusat sumber arah utara                                     |
| Kedalaman sumber $(d)$        |               3 | km      | Kedalaman di bawah permukaan                                |
| Perubahan volume $(\Delta V)$ | $(5\times10^6)$ | m³      | Sumber inflasi sintetis                                     |
| Rasio Poisson $(\nu)$         |            0,25 | –       | Parameter elastik batuan                                    |
| Sudut datang radar $(\theta)$ |              35 | derajat | Diukur dari vertikal                                        |
| Azimut pandang $(\alpha)$     |              90 | derajat | Arah horizontal pandang sensor, dari utara searah jarum jam |
| Derau                         |               0 | m       | Tidak ditambahkan agar pola dasar mudah diamati             |

Tanggal pengamatan tidak ditetapkan karena simulasi tidak menggunakan pasangan citra satelit. Parameter tersebut dapat diubah untuk menguji pengaruh kedalaman, perubahan volume sumber, dan geometri pengamatan terhadap deformasi yang dihasilkan.

## **4. Implementasi Python**

### **4.1 Mengimpor Pustaka**

Gunakan NumPy untuk perhitungan numerik dan Matplotlib untuk visualisasi.

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.dpi"] = 120
plt.rcParams["font.size"] = 10
```

### **4.2 Menentukan Parameter dan Membuat Grid**
Area pengamatan dibuat dari koordinat -5 hingga 5 km pada kedua arah horizontal. Perhitungan deformasi dilakukan menggunakan satuan meter agar konsisten dengan persamaan Mogi.

```python
wavelength = 0.056      
incidence_angle = 35     
azimuth = 90            

x0 = 0.0                
y0 = 0.0                 
depth = 3000.0          
delta_V = 5e6            
poisson = 0.25           

area_km = 5.0           
resolution_km = 0.1      

x_km = np.arange(-area_km, area_km + resolution_km / 2,
                 resolution_km)
y_km = np.arange(-area_km, area_km + resolution_km / 2,
                 resolution_km)

X_km, Y_km = np.meshgrid(x_km, y_km)

X = X_km * 1000
Y = Y_km * 1000
x0_m = x0 * 1000
y0_m = y0 * 1000

dx = X - x0_m
dy = Y - y0_m
r = np.sqrt(dx**2 + dy**2)

print("Jumlah titik grid:", X.size)
print("Ukuran grid:", X.shape)
```

### **4.3 Menghitung Deformasi Permukaan dengan Model Mogi**

Perpindahan vertikal dihitung untuk seluruh titik grid. Komponen horizontal radial kemudian diuraikan menjadi perpindahan ke timur dan utara. Pada pusat sumber, perpindahan horizontal bernilai nol karena tidak ada arah radial horizontal yang dominan.

```python
factor = (1 - poisson) * delta_V / np.pi
denominator = (r**2 + depth**2)**1.5

u_z = factor * depth / denominator
u_r = factor * r / denominator

u_E = np.zeros_like(r)
u_N = np.zeros_like(r)

mask = r > 0
u_E[mask] = u_r[mask] * dx[mask] / r[mask]
u_N[mask] = u_r[mask] * dy[mask] / r[mask]

print(f"Deformasi vertikal maksimum: {np.max(u_z)*1000:.2f} mm")
print(f"Deformasi vertikal minimum: {np.min(u_z)*1000:.2f} mm")
print(f"Perpindahan horizontal maksimum: {np.max(u_r)*1000:.2f} mm")
```

### **4.4 Visualisasi 1 — Model Deformasi Permukaan**

Peta ini menunjukkan pola deformasi vertikal sintetis yang dihasilkan oleh sumber inflasi. Nilai positif menunjukkan pengangkatan permukaan.

```python
fig, ax = plt.subplots(figsize=(8, 6))

im = ax.pcolormesh(
    X_km, Y_km, u_z * 1000,
    shading="auto",
    cmap="RdBu_r")

ax.scatter(x0, y0, color="black", marker="x",
           s=70, label="Pusat sumber")
ax.set_title("Model Deformasi Vertikal Sintetis (Mogi)")
ax.set_xlabel("Jarak Timur-Barat (km)")
ax.set_ylabel("Jarak Utara-Selatan (km)")
ax.set_aspect("equal")
ax.legend()

cbar = plt.colorbar(im, ax=ax)
cbar.set_label("Perpindahan vertikal (mm)")

plt.tight_layout()
plt.show()
```

**Interpretasi:** Pola pengangkatan terbesar berada di sekitar pusat sumber, kemudian berkurang secara bertahap menjauhi pusat tersebut. Bentuk yang relatif simetris muncul karena sumber dimodelkan berada di bawah pusat area dan medium diasumsikan homogen serta isotropik.

### **4.5 Menghitung Deformasi LOS**

Deformasi LOS diperoleh dengan memproyeksikan komponen perpindahan timur, utara, dan vertikal terhadap arah pandang radar. Dalam simulasi ini, azimut 90° berarti arah horizontal pandang sensor menuju permukaan mengarah ke timur.

```python
theta = np.deg2rad(incidence_angle)
alpha = np.deg2rad(azimuth)

d_LOS = (
    -u_E * np.sin(theta) * np.sin(alpha)
    -u_N * np.sin(theta) * np.cos(alpha)
    +u_z * np.cos(theta)
)

print(f"LOS maksimum: {np.max(d_LOS)*1000:.2f} mm")
print(f"LOS minimum: {np.min(d_LOS)*1000:.2f} mm")
```

### **4.6 Visualisasi 2 — Peta Deformasi LOS**

```python
fig, ax = plt.subplots(figsize=(8, 6))

im = ax.pcolormesh(
    X_km, Y_km, d_LOS * 1000,
    shading="auto",
    cmap="RdBu_r"
)

ax.scatter(x0, y0, color="black", marker="x",
           s=70, label="Pusat sumber")
ax.set_title("Deformasi Line of Sight (LOS) Sintetis")
ax.set_xlabel("Jarak Timur-Barat (km)")
ax.set_ylabel("Jarak Utara-Selatan (km)")
ax.set_aspect("equal")
ax.legend()

cbar = plt.colorbar(im, ax=ax)
cbar.set_label("Perpindahan LOS (mm)")

plt.tight_layout()
plt.show()
```

**Interpretasi:** Peta LOS menunjukkan bagian deformasi yang terukur dalam arah pandang radar. Karena sumber berada di bawah pusat area dan gerakan horizontal bersifat radial, komponen horizontal dapat membuat pola LOS tidak sepenuhnya simetris. Untuk geometri yang digunakan, komponen horizontal arah timur-barat memengaruhi nilai LOS secara langsung.

### **4.7 Menghitung Fase Interferometrik dan Fase Terbungkus**

Perubahan fase dihitung dari deformasi LOS. Fase terbungkus kemudian dibatasi ke rentang $([-\pi,\pi])$ menggunakan fungsi sinus dan kosinus.

```python
# Fase akibat deformasi (radian)
phase_def = -4 * np.pi / wavelength * d_LOS

# Fase terbungkus dalam rentang [-pi, pi]
phase_wrapped = np.arctan2(
    np.sin(phase_def),
    np.cos(phase_def))

print(f"Fase minimum sebelum wrapping: {np.min(phase_def):.2f} rad")
print(f"Fase maksimum sebelum wrapping: {np.max(phase_def):.2f} rad")
print(f"Fase minimum terbungkus: {np.min(phase_wrapped):.2f} rad")
print(f"Fase maksimum terbungkus: {np.max(phase_wrapped):.2f} rad")
```

### **4.8 Visualisasi 3 — Interferogram Sintetis**

```python
fig, ax = plt.subplots(figsize=(8, 6))

im = ax.pcolormesh(
    X_km, Y_km, phase_wrapped,
    shading="auto",
    cmap="twilight",
    vmin=-np.pi,
    vmax=np.pi
)

ax.scatter(x0, y0, color="white", marker="x",
           s=70, label="Pusat sumber")
ax.set_title("Interferogram Sintetis — Fase Terbungkus")
ax.set_xlabel("Jarak Timur-Barat (km)")
ax.set_ylabel("Jarak Utara-Selatan (km)")
ax.set_aspect("equal")
ax.legend()

cbar = plt.colorbar(im, ax=ax, ticks=[-np.pi, 0, np.pi])
cbar.ax.set_yticklabels([r"$-\pi$", "0", r"$+\pi$"])
cbar.set_label("Fase terbungkus (radian)")

plt.tight_layout()
plt.show()
```

**Interpretasi:** Interferogram memperlihatkan perubahan warna yang berulang akibat fase yang dibatasi ke rentang $([-\pi,\pi])$. Siklus warna bukan berarti permukaan bergerak bolak-balik secara berulang, melainkan menunjukkan ambiguitas fase akibat pembungkusan. Interpretasi deformasi dari fase terbungkus saja tidak selalu unik.

### **4.9 Profil Melintang Deformasi**

Profil melalui bagian tengah area dapat digunakan untuk membandingkan deformasi vertikal dengan deformasi LOS.

```python
iy = np.argmin(np.abs(y_km - y0))

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(
    x_km, u_z[iy] * 1000,
    label="Deformasi vertikal",
    linewidth=2
)

ax.plot(
    x_km, d_LOS[iy] * 1000,
    label="Deformasi LOS",
    linewidth=2
)

ax.axhline(0, color="black", linewidth=0.8)
ax.set_title("Profil Melintang Deformasi Sintetis")
ax.set_xlabel("Jarak horizontal (km)")
ax.set_ylabel("Perpindahan (mm)")
ax.grid(alpha=0.3)
ax.legend()

plt.tight_layout()
plt.show()
```


**Interpretasi:** Profil memperlihatkan perbedaan antara perpindahan vertikal dan perpindahan LOS. Deformasi vertikal menggambarkan gerakan naik-turun permukaan, sedangkan LOS merupakan proyeksi perpindahan tiga dimensi ke arah pandang satelit. Perbedaan keduanya bergantung pada geometri pengamatan dan komponen horizontal perpindahan.

Bobot harus dinormalisasi sehingga:

$$

Longsor merupakan fenomena geologi berupa pergerakan massa tanah atau batuan menuruni lereng akibat pengaruh gravitasi dan kondisi lingkungan. Kejadian longsor dapat dipengaruhi oleh kemiringan lereng, curah hujan, karakteristik batuan, penggunaan lahan, serta kondisi drainase dan aktivitas manusia. Sistem Informasi Geografis (SIG) dapat digunakan untuk mengintegrasikan berbagai parameter tersebut dalam bentuk peta. Python mendukung proses analisis melalui pengolahan data numerik, reklasifikasi parameter, pembobotan, dan visualisasi spasial.

Proyek ini bertujuan menghasilkan model kerawanan longsor menggunakan metode *Weighted Linear Combination* (WLC). Analisis dilakukan dengan menggabungkan skor kemiringan lereng, curah hujan, litologi, dan penggunaan lahan untuk menghasilkan peta kerawanan dengan lima kelas.

## **2. Dasar Teori**

### **2.1 Kerawanan Longsor**

Kerawanan longsor menunjukkan kecenderungan suatu wilayah untuk mengalami longsor berdasarkan karakteristik fisik dan lingkungannya. Wilayah dengan lereng curam, curah hujan tinggi, batuan yang relatif lemah, atau tutupan lahan yang kurang melindungi tanah dapat memiliki kecenderungan kerawanan yang lebih tinggi.

Namun, pengaruh setiap parameter bergantung pada kondisi geologi dan lingkungan setempat. Oleh karena itu, skor yang digunakan dalam model harus dipahami sebagai asumsi analisis, bukan ukuran universal.

### 2.2 Kemiringan Lereng

Kemiringan lereng dapat dihitung dari Digital Elevation Model (DEM), yaitu data elevasi permukaan bumi.

$$
\theta=\tan^{-1}\left(
\sqrt{
\left(\frac{\partial z}{\partial x}\right)^2+
\left(\frac{\partial z}{\partial y}\right)^2
}
\right)
$$

Keterangan:

* $(\theta)$ = kemiringan lereng dalam derajat.
* $(z)$ = elevasi permukaan.
* $(x,y)$ = koordinat horizontal dalam satuan jarak yang sama dengan elevasi.

Nilai kemiringan yang lebih besar akan diberi skor lebih tinggi pada model ini. Perhitungan menggunakan jarak horizontal dalam meter agar hasil kemiringan tidak salah akibat penggunaan koordinat geografis dalam derajat.

### **2.3 Metode Weighted Linear Combination**

WLC merupakan metode pengambilan keputusan spasial dengan menggabungkan beberapa parameter yang telah diberi skor dan bobot.

$$
S(x,y)=\sum_{i=1}^{n}w_i r_i(x,y)
$$

Bobot harus dinormalisasi sehingga:

$$
\sum_{i=1}^{n}w_i=1
$$

Skor akhir yang lebih tinggi menunjukkan kerawanan relatif lebih tinggi berdasarkan asumsi model. Metode ini bersifat kompensatoris, artinya skor rendah pada satu parameter dapat diimbangi skor tinggi pada parameter lainnya.

## **4. Implementasi Python**

### **4.1 Mengimpor Pustaka**

Pustaka NumPy digunakan untuk perhitungan numerik, Pandas untuk menyusun tabel, dan Matplotlib untuk membuat visualisasi.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm

plt.rcParams["figure.dpi"] = 120
plt.rcParams["font.size"] = 10

# Seed agar simulasi dapat direproduksi
rng = np.random.default_rng(42)
```

### **4.2 Membuat Grid Wilayah dan DEM Sintetis**

DEM sintetis dibuat dari beberapa bentuk bukit dan punggungan. Bentuk permukaan ini sengaja dibuat bervariasi agar menghasilkan kemiringan lereng yang berbeda.

```python
cell_size = 100      
n = 100               
area_km = 10         

coord_km = (np.arange(n) + 0.5) * cell_size / 1000 - area_km / 2
X_km, Y_km = np.meshgrid(coord_km, coord_km)

X = X_km * 1000
Y = Y_km * 1000

DEM = (
    350
    + 650 * np.exp(
        -(((X_km + 1.2) / 1.4)**2 + ((Y_km - 0.8) / 1.8)**2)
    )
    + 500 * np.exp(
        -(((X_km - 2.0) / 0.8)**2 + ((Y_km + 1.5) / 1.5)**2)
    )
    + 250 * np.exp(-((X_km + Y_km) / 0.9)**2)
)

DEM += rng.normal(0, 5, size=DEM.shape)

print("Ukuran raster:", DEM.shape)
print(f"Elevasi minimum: {DEM.min():.2f} m")
print(f"Elevasi maksimum: {DEM.max():.2f} m")
```

### **4.3 Menghitung Kemiringan Lereng**

Kemiringan lereng dihitung dari perubahan elevasi pada arah horizontal. Fungsi `np.gradient` menerima ukuran piksel dalam meter sehingga gradien elevasi memiliki satuan yang konsisten.

```python
# Gradien elevasi terhadap arah utara-selatan dan timur-barat
dz_dy, dz_dx = np.gradient(DEM, cell_size, cell_size)

# Kemiringan dalam derajat
slope = np.degrees(
    np.arctan(np.sqrt(dz_dx**2 + dz_dy**2)))

print(f"Kemiringan minimum: {slope.min():.2f}°")
print(f"Kemiringan maksimum: {slope.max():.2f}°")
```

### **4.4 Membuat Parameter Curah Hujan, Litologi, dan Penggunaan Lahan**

Karena tidak menggunakan data observasi, ketiga parameter dibuat sebagai pola spasial sintetis. Nilai ini hanya menggambarkan variasi antarlokasi untuk menguji alur analisis.

```python
# Curah hujan sintetis (mm/tahun)
rainfall = (
    1900
    + 180 * (Y_km / 5)
    + 650 * np.exp(
        -(((X_km - 2.0) / 2.0)**2 + ((Y_km - 2.0) / 2.0)**2)
    )
    + rng.normal(0, 45, size=(n, n))
)

# Pastikan nilai hujan tidak negatif
rainfall = np.clip(rainfall, 500, None)

# Litologi sintetis:
# 1 = batuan relatif kompak
# 2 = batuan berkekuatan menengah
# 3 = material relatif lemah/lapuk
lithology = np.ones((n, n), dtype=int)

lithology[(X_km > -1.5) & (Y_km < 1.5)] = 2
lithology[(X_km > 1.0) & (Y_km > 0.5)] = 3
lithology[(X_km < -2.0) & (Y_km < -1.0)] = 3

# Penggunaan lahan sintetis:
# 1 = hutan
# 2 = pertanian
# 3 = terbangun
# 4 = lahan terbuka
landcover = np.ones((n, n), dtype=int)

landcover[(Y_km < 0.5) & (X_km > -2.0)] = 2
landcover[(X_km < -2.0) & (Y_km < 1.0)] = 3
landcover[(X_km > 1.5) & (Y_km > 1.0)] = 4

# Tampilkan rentang parameter
print(
    f"Curah hujan sintetis: {rainfall.min():.0f}–"
    f"{rainfall.max():.0f} mm/tahun"
)
print("Kelas litologi:", np.unique(lithology))
print("Kelas penggunaan lahan:", np.unique(landcover))
```

### **4.5 Melakukan Reklasifikasi Parameter**

Setiap parameter diubah menjadi skor 1–5. Skor yang lebih tinggi menunjukkan kontribusi kerawanan relatif lebih tinggi berdasarkan asumsi yang digunakan.

```python
# Reklasifikasi kemiringan lereng
slope_score = np.digitize(
    slope,
    bins=[8, 15, 25, 40]
) + 1

# Reklasifikasi curah hujan
rain_score = np.digitize(
    rainfall,
    bins=[1500, 2000, 2500, 3000]
) + 1

# Reklasifikasi litologi
lithology_score = np.select(
    [
        lithology == 1,
        lithology == 2,
        lithology == 3
    ],
    [1, 3, 5]
)

# Reklasifikasi penggunaan lahan
landcover_score = np.select(
    [
        landcover == 1,
        landcover == 2,
        landcover == 3,
        landcover == 4
    ],
    [1, 3, 4, 5]
)

print("Skor lereng:", np.unique(slope_score))
print("Skor hujan:", np.unique(rain_score))
print("Skor litologi:", np.unique(lithology_score))
print("Skor penggunaan lahan:", np.unique(landcover_score))
```

### **4.6 Menampilkan Peta Kemiringan Lereng**

Peta pertama menunjukkan distribusi kemiringan lereng hasil perhitungan dari DEM sintetis.

```python
fig, ax = plt.subplots(figsize=(8, 6))

im = ax.pcolormesh(
    X_km, Y_km, slope,
    shading="auto",
    cmap="terrain"
)

ax.set_title("Peta Kemiringan Lereng Sintetis")
ax.set_xlabel("Koordinat lokal arah timur-barat (km)")
ax.set_ylabel("Koordinat lokal arah utara-selatan (km)")
ax.set_aspect("equal")

cbar = plt.colorbar(im, ax=ax)
cbar.set_label("Kemiringan lereng (derajat)")

plt.tight_layout()
plt.show()
```

**Interpretasi:** Area dengan warna yang menunjukkan kemiringan lebih besar merupakan bagian dengan lereng lebih curam. Dalam model ini, lereng curam akan mendapatkan skor lebih tinggi. Pola aktual dan rentang nilainya perlu diperiksa melalui hasil kode, bukan diasumsikan sebelumnya.

### **4.7 Menampilkan Peta Setiap Parameter**

Peta skor digunakan untuk melihat bagaimana masing-masing parameter memberikan kontribusi terhadap model WLC.

```python
parameter_maps = [
    (slope_score, "Skor Kemiringan Lereng"),
    (rain_score, "Skor Curah Hujan"),
    (lithology_score, "Skor Litologi"),
    (landcover_score, "Skor Penggunaan Lahan")]

score_cmap = ListedColormap([
    "#2c7bb6",
    "#abd9e9",
    "#ffffbf",
    "#fdae61",
    "#d7191c"])
score_norm = BoundaryNorm(
    np.arange(0.5, 6, 1),
    score_cmap.N)

fig, axes = plt.subplots(2, 2, figsize=(12, 9))

for ax, (data, title) in zip(axes.flat, parameter_maps):
    im = ax.pcolormesh(
        X_km, Y_km, data,
        shading="auto",
        cmap=score_cmap,
        norm=score_norm)

    ax.set_title(title)
    ax.set_xlabel("Timur-barat (km)")
    ax.set_ylabel("Utara-selatan (km)")
    ax.set_aspect("equal")

fig.colorbar(
    im,
    ax=axes.ravel().tolist(),
    ticks=[1, 2, 3, 4, 5],
    label="Skor parameter",
    shrink=0.8)

fig.suptitle("Peta Skor Parameter Kerawanan Longsor Sintetis")
plt.show()
```

**Interpretasi:** Keempat peta memperlihatkan variasi spasial skor masing-masing parameter. Peta ini membantu menelusuri alasan suatu lokasi memperoleh skor kerawanan tertentu. Seluruh pola merupakan hasil pembentukan data sintetis.

### **4.8 Menghitung Skor Kerawanan Menggunakan WLC**

Skor akhir diperoleh dari penjumlahan skor setiap parameter yang dikalikan dengan bobotnya.

```python
# Bobot parameter
weights = {
    "slope": 0.35,
    "rainfall": 0.25,
    "lithology": 0.25,
    "landcover": 0.15
}

# Pemeriksaan normalisasi bobot
assert np.isclose(sum(weights.values()), 1.0), \
    "Jumlah bobot harus sama dengan 1."

# Skor kerawanan WLC
susceptibility = (
    weights["slope"] * slope_score
    + weights["rainfall"] * rain_score
    + weights["lithology"] * lithology_score
    + weights["landcover"] * landcover_score
)

print(f"Skor minimum WLC: {susceptibility.min():.2f}")
print(f"Skor maksimum WLC: {susceptibility.max():.2f}")
print(f"Bobot total: {sum(weights.values()):.2f}")
```

### **4.9 Mengklasifikasikan Tingkat Kerawanan**

Skor WLC memiliki rentang teoritis 1–5 karena semua skor parameter berada dalam rentang tersebut dan bobot berjumlah satu. Untuk simulasi ini digunakan interval sama dengan lebar 0,8.

| Skor WLC | Kelas         |
| -------- | ------------- |
| 1,0–1,8  | Sangat rendah |
| >1,8–2,6 | Rendah        |
| >2,6–3,4 | Sedang        |
| >3,4–4,2 | Tinggi        |
| >4,2–5,0 | Sangat tinggi |

Batas kelas tersebut merupakan pilihan klasifikasi untuk demonstrasi. Metode klasifikasi lain dapat menghasilkan distribusi kelas yang berbeda.

```python
class_bounds = [1.8, 2.6, 3.4, 4.2]

susceptibility_class = np.digitize(
    susceptibility,
    bins=class_bounds
) + 1

class_names = [
    "Sangat rendah",
    "Rendah",
    "Sedang",
    "Tinggi",
    "Sangat tinggi"]

print("Kelas yang muncul:", np.unique(susceptibility_class))
```

### **4.10 Visualisasi Peta Kerawanan Longsor**
Peta akhir menunjukkan kelas kerawanan relatif yang dihasilkan dari integrasi empat parameter.

```python
susceptibility_cmap = ListedColormap([
    "#2c7bb6",  # Sangat rendah
    "#abd9e9",  # Rendah
    "#ffffbf",  # Sedang
    "#fdae61",  # Tinggi
    "#d7191c"]) # Sangat tinggi])

class_norm = BoundaryNorm(
    np.arange(0.5, 6, 1),
    susceptibility_cmap.N)

fig, ax = plt.subplots(figsize=(9, 7))

im = ax.pcolormesh(
    X_km, Y_km, susceptibility_class,
    shading="auto",
    cmap=susceptibility_cmap,
    norm=class_norm)

ax.set_title("Peta Kerawanan Longsor Sintetis — Metode WLC")
ax.set_xlabel("Koordinat lokal arah timur-barat (km)")
ax.set_ylabel("Koordinat lokal arah utara-selatan (km)")
ax.set_aspect("equal")

# Panah utara
ax.annotate(
    "N",
    xy=(0.93, 0.91),
    xytext=(0.93, 0.79),
    xycoords="axes fraction",
    ha="center",
    arrowprops=dict(arrowstyle="-|>", color="black", lw=1.5))

# Skala grafis 2 km
ax.plot(
    [-4.5, -2.5], [-4.5, -4.5],
    color="black", linewidth=3)
ax.text(-3.5, -4.3, "2 km", ha="center")

cbar = plt.colorbar(
    im,
    ax=ax,
    ticks=[1, 2, 3, 4, 5],
    boundaries=np.arange(0.5, 6, 1))
cbar.ax.set_yticklabels(class_names)
cbar.set_label("Kelas kerawanan")

plt.tight_layout()
plt.show()
```

**Interpretasi:** Peta memperlihatkan distribusi kelas kerawanan berdasarkan kombinasi skor parameter. Area dengan kelas tinggi atau sangat tinggi memperoleh skor WLC lebih besar daripada area dengan kelas rendah. Lokasi spesifik yang termasuk kelas tertentu harus diidentifikasi langsung dari peta yang dihasilkan.

Panah utara dan skala grafis merupakan elemen orientasi pada grid lokal simulasi. Peta ini bukan peta berkoordinat geografis dan tidak memiliki batas administrasi wilayah nyata.

### **4.11 Menghitung Luas dan Persentase Setiap Kelas**

Karena ukuran piksel adalah 100 × 100 m, luas satu piksel adalah 10.000 m² atau 1 hektare. Luas setiap kelas dapat dihitung berdasarkan jumlah piksel yang masuk ke kelas tersebut.

```python
pixel_area_ha = (cell_size * cell_size) / 10000

total_cells = susceptibility_class.size
total_area_ha = total_cells * pixel_area_ha

area_results = []

for class_id, class_name in enumerate(class_names, start=1):
    count = np.count_nonzero(susceptibility_class == class_id)
    area_ha = count * pixel_area_ha
    percentage = count / total_cells * 100

    area_results.append({
        "Kelas kerawanan": class_name,
        "Jumlah piksel": count,
        "Luas (ha)": area_ha,
        "Luas (km²)": area_ha / 100,
        "Persentase (%)": percentage
    })

area_table = pd.DataFrame(area_results)

print(f"Luas total simulasi: {total_area_ha:.2f} ha")
display(area_table.round(2))
```

Tabel tersebut menunjukkan luas dan persentase kelas berdasarkan hasil perhitungan aktual dari kode. Kelas yang tidak muncul pada simulasi akan memiliki luas nol.

### **4.12 Visualisasi Luas Setiap Kelas**

```python
fig, ax = plt.subplots(figsize=(9, 5))

ax.bar(
    area_table["Kelas kerawanan"],
    area_table["Luas (ha)"]
)

ax.set_title("Luas Setiap Kelas Kerawanan Longsor Sintetis")
ax.set_xlabel("Kelas kerawanan")
ax.set_ylabel("Luas (hektare)")
ax.tick_params(axis="x", rotation=20)
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

**Interpretasi:** Diagram memperlihatkan perbandingan luas antarkelas kerawanan. Kelas dengan luas terbesar merupakan kelas yang mencakup jumlah piksel paling banyak pada model sintetis ini. Hasilnya perlu dibaca bersama peta spasial karena diagram luas tidak menunjukkan lokasi distribusi setiap kelas.
